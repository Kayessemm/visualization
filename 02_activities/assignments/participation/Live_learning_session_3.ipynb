```python
# Customization live-code

import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
```

```python
# Add legend:

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
          frameon = True, # Add frame around the legend
          fontsize = 12, #change font size
          ncol = 2, #Specify number of columns
          shadow = True)
fig.show()
```

```python
# However, the legend is in the box, on top of the lines, so we'll move it out

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='upper left', # Attaches the legend's upper left side to the corner of the graph
          bbox_to_anchor =(1,1)) # Anchors legend at 1, 1 (top right) coordinates of the axis coordinate system (e.g., 0, 0 is the bottom left)
fig.show()
```

```python
# Flage a certain value

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!")
fig.show()
```

```python
# Sometimes we need the label in the same place, regardless of where data are.

# start with simple plot

fig, ax=plt.subplots()
ax.axis([0,10,0,10])

ax.text(1, 5, ". Data:(1,5)",  transform=ax.transData)
ax.text(0.5, 0.1, ".Axes:(0.5, 0.1)",   transform=ax.transAxes)
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)",    transform=fig.transFigure)

# These fractions are relevant to the whole caonvas, not to the subplots
```

```python
# We can annotate with arrows

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.annotate("This is important!", xy=(10,90), xytext=(20,98), #Space between annotation and point
            arrowprops=dict(facecolor='black'))
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', 
            xy=(10, 95), 
            xytext=(20, 94), 
            arrowprops = dict(arrowstyle = "wedge", 
                color = " hotpink "))
```

```python
# Remove tick-marks and labels

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator()) # Removes Y-axis labels and tick marks
ax.xaxis.set_major_formatter(plt.NullFormatter()) #Removes X-axis labels
```

```python
# Specify the number of tick-marks

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
#This code sets the intervals of the tick marks

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
# Rotate axis labels:

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

plt.xticks(rotation=45, ha='right')
```

```python
# My attempt to modify the x-axis title; added Y label for fun

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xlabel("Shiny New X", 
            color = 'indigo',
            font = 'serif')
plt.ylabel("Shiny New Y", 
            color = 'indigo',
            font = 'serif')
```

```python
# Actual way to add label:
font1 = {'family':'serif','color':'indigo'} #Define font family
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xlabel("Shiny New X-axis!", fontsize = 18, fontdict = font1) 
```
