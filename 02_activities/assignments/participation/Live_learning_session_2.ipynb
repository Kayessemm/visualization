```python
# 2nd visualization session

# This lecture is about how to choose the right data visualization; this will include discussion of helpful sources, particularly D'Ignazio and Klein's (2020) "Data Feminism"
# 
# To recap, a figure is a container that holds axes; axes are where data are ploted; an artist defines the appearance of these axes. 
# 
# For today: What can we learn from a data visualization?
# What is the objectivity and neutrality of the visualization? 
# 
# Gun killing figure: orange part of arch was life lived, grey was years that could have been lived called "stolen years"
    # Shows objective info about when people died and how, but the estimates of potential lifespans/stolen years are highly conjectural, and meant to emphasize a point
    # Whether it's a good visualization depends on the purpose: e.g., is it meant to elicit an emotional response?  
# Era of "active shooters" figure
    # A bar graph that simply shows the amount of shooters per year, objectively showing increase over time.
    # Muted colours were also used to minimize emotional response
# Same "Blank page" minimalism used for unemployment rate figure from NYT, except one emphasizes how unemployment rate stays above 8% (negative framing) versus a steady decline (positive framing)
# Can visualization be neutral?
    # No; choices inherently inform how we present facts through visualizations
    # Still, there are four basic rules for maximizing objectivity (discussed later)   
# D'ignazio and Klein (2020) emphaize that while visualizations can be factual, they may not be neutral, especially given how they are used in *rhetoric*
# Four purposes of visualization:
    # Persuasion
    # Comparison
    # Evaluation
    # Exploration 
# Audience also important:
    # Age
    # Education
    # Expertise
    # Accessibility 
# Also consider intended medium:
    # Print (e.g., should be more legible/simpler; perhaps greyscale)
    # Web
    # Poster
    # Presentation 
# There are also different dataset types:
    # Table
        # Often best with bar charts, scatterplots
    # Networks
        # Relationships most important to represent
    # Fields
        # Values on a grid (e.g., heatmap, location plot)
    # Geometry (i.e., maps) 
# Attribute types help determine how to visualize them
    # Categorical data may have different symbols
    # Ordered (i.e., ordinal or quantitative) may be symbolized by size
    # Ordering directions are usually represented with arrows or colour gradients  
# How do people perceive and understand visualizations?
    # Gestalt principles
        # Proximity defines groups
        # Similarity also defines groups
        # Continuity
        # Closure: brain automatically aims to close gaps
        # Enclosure: objects in a boundary are perceived as the same group
        # Connection links items into groups
# Cognitive load:
    # Intrinsic (how complex the info is)
    # Germane (audience's familiarity with content)
    # Extraneous (complexity of how info is presented)
# Cognitive load can be affected by:
    # Familiar vs. rare chart types
    # Accurate vs. approximate interpretation
    # Concise vs. detailed composition
    # Explanatory vs. exploratory composition (more or less guidance for audience)
# Cognition over memory: better to show side-by-side figures for comparison
# 3D format is usually not recommended unless necessary, because it increases cognitive load
# Provenience rhetoric: better to include source in/with figure to increase objectivity
# datavizcatalogue.com
# python-graph-gallery.com
```
